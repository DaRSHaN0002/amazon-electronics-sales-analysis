```python
df = df.drop_duplicates()
```

```python
df.describe()
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
```

```python
df["Order_Date"].dtype
```

```python
df["Customer_Rating"] = df["Customer_Rating"].fillna(
    df["Customer_Rating"].median()
)
```

```python
df["Payment_Method"] = df["Payment_Method"].fillna(
    df["Payment_Method"].mode()[0]
)
```

```python
category_sales = (
    df.groupby("Category")["Sales_INR"]
    .sum()
    .sort_values(ascending=False)
)

category_sales
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=category_sales.index,
    y=category_sales.values
)

plt.title("Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Sales (INR)")
plt.xticks(rotation=30)

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=category_sales.index,
    y=category_sales.values
)

plt.title("Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Sales (INR)")
plt.xticks(rotation=30)

plt.tight_layout()

plt.savefig("../visuals/sales_by_category.png", dpi=300)

plt.show()
```

```python
monthly_sales = (
    df.groupby("Month_Number")["Sales_INR"]
    .sum()
    .sort_index()
)

monthly_sales
```

```python
plt.figure(figsize=(12, 6))

sns.lineplot(
    x=monthly_sales.index,
    y=monthly_sales.values,
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales (INR)")
plt.xticks(range(1, 13))

plt.tight_layout()

plt.savefig("../visuals/monthly_sales_trend.png", dpi=300)

plt.show()
```

```python
best_month = monthly_sales.idxmax()
best_month_sales = monthly_sales.max()

worst_month = monthly_sales.idxmin()
worst_month_sales = monthly_sales.min()

print("Best month:", best_month)
print("Best month sales:", best_month_sales)

print("Worst month:", worst_month)
print("Worst month sales:", worst_month_sales)
```

```python
month_names = [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
]
```

```python
plt.figure(figsize=(12, 6))

sns.lineplot(
    x=month_names,
    y=monthly_sales.values,
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales (INR)")

plt.tight_layout()

plt.savefig("../visuals/monthly_sales_trend.png", dpi=300)

plt.show()
```

```python
average_order_value = df["Sales_INR"].mean()

print("Average Order Value:", average_order_value)
```

```python
average_quantity = df["Quantity"].mean()

print("Average Quantity per Order:", average_quantity)
```

```python
category_summary = (
    df.groupby("Category")
    .agg(
        Total_Sales=("Sales_INR", "sum"),
        Total_Quantity=("Quantity", "sum"),
        Average_Rating=("Customer_Rating", "mean"),
        Average_Discount=("Discount_Percent", "mean"),
        Number_of_Orders=("Order_ID", "nunique")
    )
    .sort_values("Total_Sales", ascending=False)
)

category_summary
```

```python
category_summary.to_excel(
    "../data/category_performance_summary.xlsx"
)
```
